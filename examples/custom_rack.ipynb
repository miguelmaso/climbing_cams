```python
import climbing_cams
import matplotlib.pyplot as plt
```

```python
rack = climbing_cams.db.select(name="Z4", range=[0,20])
rack += climbing_cams.db.select(brand="Totem", range=[10,60])
rack.barh(ylabel='{name} [{number}]')
plt.text(.9, .1, f'Total weight: {rack.weight} {climbing_cams.units.Measurements.weight.label}',
         transform = plt.gca().transAxes, horizontalalignment='right',
         bbox=dict(facecolor='none', edgecolor='black', boxstyle='round,pad=1'))
plt.tight_layout()
plt.show()
```
