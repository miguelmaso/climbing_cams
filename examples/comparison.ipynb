```python
import climbing_cams as cams
import matplotlib.pyplot as plt
```

```python
family_specifications = [
    {'brand':'Metolius',      'name':'Master Cam'},
    {'brand':'Totem',         'name':'Cam'},
    {'brand':'Alien',         'name':'X'},
    {'brand':'Black Diamond', 'name':'Z4'},
    {'brand':'Black Diamond', 'name':'C4'},
    {'brand':'DMM',           'name':'Dragon'},
    {'brand':'Wild Country',  'name':'Friend'}
]
families = cams.db.select_many(family_specifications)
```

```python
families.barh()
plt.gcf().set_size_inches(10,7)
plt.show()
```

```python
families.scatter('expansion_rate', 'specific_weight')
plt.show()
```

```python
fig, ax = plt.subplots(2,1,figsize=(10,5))
families.scatter('avg', 'weight', ax=ax[0], individual=True)
families.scatter('avg', 'specific_weight', ax=ax[1], individual=True)
ax[0].set_yscale('log')
ax[0].set_xscale('log')
ax[1].set_yscale('log')
ax[1].set_xscale('log')
ax[0].get_legend().remove()
plt.show()
```
